### Task 1.2: Problem Statement
**Objective:** Define customer churn in telecommunications and explain why predictive classification is essential.
**Method:** Formulate the mathematical and operational characteristics of subscription turnover.

In subscription-based telecommunications, subscribers generate monthly recurring revenue. When a subscriber disconnects service or switches to a competitor, the provider forfeits future recurring cash flows and incurs heavy marketing costs to acquire a replacement subscriber. 

Predicting churn is framed as a supervised binary classification problem:
$$y_i \\in \\{0, 1\\}, \\quad \\text{where } 1 = \\text{Customer Churned (Yes)}, \\, 0 = \\text{Customer Retained (No)}$$
Given an input feature vector $\\mathbf{x}_i \\in \\mathbb{R}^p$ representing customer demographics, subscription tenure, subscribed services, contract commitment, and billing configurations, our goal is to construct an optimal decision function $\\hat{f}(\\mathbf{x}_i)$ that estimates the posterior class probability:
$$\\hat{p}_i = P(y_i = 1 \\mid \\mathbf{x}_i)$$
A major analytical challenge is that customer churn datasets typically exhibit **class imbalance**, where non-churners outnumber churners. Evaluation must therefore emphasize metric balance—specifically prioritizing the harmonic mean of precision and recall (**F1-score**) alongside the Area Under the ROC Curve (**ROC-AUC**).

**Defensible Imputation Strategy:**
Computing a global mean or median across the entire dataset before train-test splitting introduces **data leakage** (contaminating the training process with future test information). 3. Fit the imputer **strictly on training fold data** and transform both training and test folds downstream.

```python
# Task 3.3: Handle Missing TotalCharges
print(f"Total rows retained in df_clean: {len(df_clean)}")
print(f"Missing TotalCharges retained for pipeline imputation: {df_clean['TotalCharges'].isna().sum()}")
display(df_clean["TotalCharges"].describe())
```

### Task 4.11: Identify Outliers
**Objective:** Detect and evaluate potential statistical outliers in numerical variables using boxplots and the Interquartile Range (IQR) rule.
**Method:** Calculate IQR boundaries and identify potential outliers for `tenure`, `MonthlyCharges`, and `TotalCharges`.

```python
# Task 4.11: Identify Outliers via IQR Rule
num_cols_for_outliers = ["tenure", "MonthlyCharges", "TotalCharges"]
outlier_summary = []

for col in num_cols_for_outliers:
    series = df_clean[col].dropna()
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers_count = ((series < lower_bound) | (series > upper_bound)).sum()
    outlier_summary.append({
        "Feature": col,
        "Q1 (25%)": round(q1, 2),
        "Q3 (75%)": round(q3, 2),
        "IQR": round(iqr, 2),
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": outliers_count,
        "Outlier Percentage (%)": round((outliers_count / len(series)) * 100, 2)
    })

display(pd.DataFrame(outlier_summary))
```

**Observation:**
Under the standard $1.5 \\times \\text{IQR}$ rule, **zero outliers** are detected in `tenure`, `MonthlyCharges`, and `TotalCharges`. All numerical values fall entirely within expected boundaries.

**Conclusion:**
The continuous features are free of statistical outliers, requiring no truncation, winsorization, or record filtering.

**Objective:** Build a reusable numerical transformation pipeline with imputation and standardization.
**Method:** Combine `SimpleImputer(strategy='median')` and `StandardScaler()` in a `Pipeline`.

```python
# Task 5.5: Define Numerical Preprocessing Pipeline
num_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
print("Numerical preprocessing pipeline created: SimpleImputer(median) -> StandardScaler().")
```

**Observation:**
The numerical transformer imputes missing values using the median (robust against skewness) and standardizes features to zero mean and unit variance ($z = \\frac{x - \\mu}{\\sigma}$).

### Task 5.6: Define Categorical Preprocessing
**Objective:** Build a categorical transformation pipeline with mode imputation and one-hot encoding.
**Method:** Combine `SimpleImputer(strategy='most_frequent')` and `OneHotEncoder(handle_unknown='ignore', sparse_output=False)`.

```python
# Task 5.6: Define Categorical Preprocessing Pipeline
cat_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])
print("Categorical preprocessing pipeline created: SimpleImputer(most_frequent) -> OneHotEncoder(handle_unknown='ignore').")
```

**Observation:**

### Task 5.7: Build ColumnTransformer

```python
# Task 5.7: Build ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_pipeline, numerical_features),
        ("cat", cat_pipeline, categorical_features)
    ],
    remainder="drop"
)
print("ColumnTransformer successfully constructed.")
```

**Observation:**

### Task 6.2: Logistic Regression
**Objective:** Build and configure a regularized linear classification model.
**Method:** Initialize `LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)` inside the preprocessing pipeline.

**Mathematical Formulation:**
Logistic Regression models the log-odds of the positive class ($y=1$) as an affine combination of input features:
$$\\log \\left( \\frac{p_i}{1 - p_i} \\right) = \\beta_0 + \\sum_{j=1}^p \\beta_j x_{ij} = \\mathbf{x}_i^T \\boldsymbol{\\beta}$$
The estimated probability of churn is obtained via the logistic sigmoid link function:
$$p_i = \\sigma(\\mathbf{x}_i^T \\boldsymbol{\\beta}) = \\frac{1}{1 + e^{-\\mathbf{x}_i^T \\boldsymbol{\\beta}}}$$
**Strengths:** Highly interpretable, provides well-calibrated probabilities, computationally efficient, convex optimization guarantees global minimum.
**Limitations:** Assumes linear decision boundary in log-odds space; cannot naturally model complex nonlinear feature interactions without explicit expansion.

```python
# Task 6.2: Logistic Regression Setup
lr_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
])
print("LogisticRegression pipeline initialized (max_iter=1000, random_state=42).")
```

**Conclusion:**

### Task 6.3: Decision Tree Classifier
**Objective:** Build and configure a non-parametric decision tree model.
**Method:** Initialize `DecisionTreeClassifier(random_state=RANDOM_STATE)` inside the preprocessing pipeline.

**Mathematical Formulation:**
A Decision Tree recursively partitions feature space into axis-aligned rectangular regions $R_m$. At each node, the split $(j, s)$ is chosen to maximize the reduction in Gini impurity:
$$I_G(t) = 1 - \\sum_{k=0}^1 p_k(t)^2$$
**Strengths:** Completely non-parametric, naturally captures nonlinearities and high-order feature interactions, invariant to monotonic feature scaling, intuitive tree rules.
**Limitations:** High variance; unconstrained trees tend to overfit training data by creating complex, brittle decision boundaries.

```python
# Task 6.3: Decision Tree Classifier Setup
dt_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(random_state=RANDOM_STATE))
])
print("DecisionTreeClassifier pipeline initialized (random_state=42).")
```

### Task 6.4: Random Forest Classifier
**Objective:** Build and configure an ensemble Random Forest model combining multiple de-correlated decision trees.
**Method:** Initialize `RandomForestClassifier(random_state=RANDOM_STATE)` inside the preprocessing pipeline.

**Mathematical Formulation:**
Random Forest reduces the high variance of individual decision trees through **Bootstrap Aggregation (Bagging)** and **Random Subspace Projection**:
$$\\hat{f}_{RF}(\\mathbf{x}) = \\frac{1}{B} \\sum_{b=1}^B T_b(\\mathbf{x}; \\Theta_b)$$
At each candidate split, only a random subset of features $m \\approx \\sqrt{p}$ is considered. This de-correlates the ensemble trees, dramatically shrinking prediction variance:
$$\\text{Var}(\\bar{T}) = \\rho \\sigma^2 + \\frac{1-\\rho}{B} \\sigma^2$$
**Strengths:** Exceptional generalization accuracy, robust against overfitting, handles nonlinearities and interactions, provides feature importances.
**Limitations:** Computationally heavier than single trees or linear models, reduced direct interpretability relative to a single decision tree.

```python
# Task 6.4: Random Forest Classifier Setup
rf_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(random_state=RANDOM_STATE))
])
print("RandomForestClassifier pipeline initialized (random_state=42).")
```

**Conclusion:**

## Introduction

## Objectives

### Task 7.1: Stratified Cross-Validation
**Objective:** Define a 5-fold Stratified Cross-Validation scheme and justify its necessity for imbalanced churn data.
**Method:** Initialize `StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)`.

**Why Stratification is Essential:**
Standard $K$-fold cross-validation randomly assigns records to folds, which risks uneven minority class distribution across validation splits. Stratified $K$-fold enforces that every fold contains approximately **26.5% churners and 73.5% non-churners**, ensuring that every fold faithfully reflects the true population distribution.

```python
# Task 7.1: Configure Stratified 5-Fold Cross-Validation
cv_stratified = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
print(f"StratifiedKFold initialized with 5 splits (shuffle=True, random_state={RANDOM_STATE}).")
```

**Observation:**
A 5-fold stratified cross-validation splitter is instantiated. Each fold will contain approximately 1,127 validation records with identical class proportions.

**Conclusion:**
Stratified validation eliminates fold-to-fold class distribution variance, ensuring stable, reliable metric estimation.

### Task 7.2: Cross-Validation Metrics
**Objective:** Compute out-of-fold validation scores across Accuracy, Precision, Recall, F1-score, and ROC-AUC.
**Method:** Execute `cross_validate()` across the 5 stratified folds for all four pipelines.

```python
# Task 7.2: Compute Cross-Validation Metrics
scoring_metrics = ["accuracy", "precision", "recall", "f1", "roc_auc"]

cv_raw_scores = {}

print("Executing 5-fold Stratified Cross-Validation across all 4 candidate pipelines...")
for name, pipe in model_pipelines.items():
    start_time = time.time()
    scores = cross_validate(
        pipe,
        X_train,
        y_train,
        cv=cv_stratified,
        scoring=scoring_metrics,
        n_jobs=-1
    )
    cv_raw_scores[name] = scores
    elapsed = time.time() - start_time
    print(f"  [✓] Completed 5-fold CV for {name:<22s} in {elapsed:.2f}s")
```

**Observation:**
All 20 cross-validation folds (4 models $\\times$ 5 folds) completed successfully.

**Conclusion:**
Cross-validation scores across all five core metrics are computed and ready for aggregation.

### Task 7.3: Model Comparison
**Objective:** Aggregate cross-validation scores into a structured comparison table and bar chart.
**Method:** Calculate mean and standard deviation for each metric and display formatted summary tables.

```python
# Task 7.3: Model Comparison Table and Visualizations
cv_summary_records = []

for name, scores in cv_raw_scores.items():
    record = {"Model": name}
    for metric in scoring_metrics:
        mean_val = scores[f"test_{metric}"].mean()
        std_val = scores[f"test_{metric}"].std()
        record[f"{metric.upper()} Mean"] = mean_val
        record[f"{metric.upper()} Std"] = std_val
    cv_summary_records.append(record)

cv_summary_df = pd.DataFrame(cv_summary_records)
cv_summary_display = cv_summary_df.copy()

# Format for clean academic presentation
for col in cv_summary_display.columns:
    if "Mean" in col or "Std" in col:
        cv_summary_display[col] = cv_summary_display[col].map("{:.4f}".format)

print("--- 5-Fold Stratified Cross-Validation Benchmark Results ---")
display(cv_summary_display)

# Visual comparison chart
metrics_to_plot = ["ACCURACY Mean", "PRECISION Mean", "RECALL Mean", "F1 Mean", "ROC_AUC Mean"]
plot_df = cv_summary_df.set_index("Model")[metrics_to_plot]
plot_df.columns = ["Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"]

fig, ax = plt.subplots(figsize=(12, 6))
plot_df.plot(kind="bar", ax=ax, rot=0, width=0.8)
ax.set_title("Cross-Validation Metric Comparison Across 4 Models", fontsize=14, fontweight="bold", pad=12)
ax.set_ylabel("Score", fontsize=11)
ax.set_ylim(0.0, 1.0)
ax.legend(bbox_to_anchor=(1.02, 1), loc="upper left", title="Metric")
plt.tight_layout()
plt.show()
```

### Task 7.6: Interpret Results

**Method:** Contrast the costs of False Positives versus False Negatives.

**Analytical Breakdown:**
1. **Misleading Nature of Accuracy:** The Dummy Classifier achieves **73.46% accuracy** while offering zero predictive value. In imbalanced domains, accuracy is a deceptive metric.
2. **False Negatives (Missed Churners):** A False Negative occurs when a customer likely to leave is predicted as retained. The company takes no retention action, and the customer churns, resulting in lost lifetime recurring revenue and substantial replacement acquisition costs.
3. **False Positives (Wasted Retention Offers):** A False Positive occurs when a loyal customer is flagged as an at-risk churner. The company may send retention promotions or discounts. While this incurs minor promotional overhead, the customer remains with the company.
4. **Metric Prioritization:** In telecom retention, **Recall** is critical to capture churners, while **Precision** prevents excessive promotional waste. Therefore, **F1-score** (the harmonic mean of precision and recall) serves as our primary selection criterion, supported by **ROC-AUC**.

```python
# Task 8.1: GridSearchCV for Logistic Regression
pipe_lr_tuning = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
])

param_grid_lr = {
    "classifier__C": [0.01, 0.1, 1.0, 10.0],
    "classifier__class_weight": [None, "balanced"]
}

grid_lr = GridSearchCV(
    estimator=pipe_lr_tuning,
    param_grid=param_grid_lr,
    scoring="f1",
    cv=cv_stratified,
    n_jobs=-1
)

start_time = time.time()
grid_lr.fit(X_train, y_train)
lr_tune_time = time.time() - start_time

print(f"GridSearchCV Logistic Regression completed in {lr_tune_time:.2f}s.")
print(f"Tested Parameter Combinations: {len(grid_lr.cv_results_['params'])}")
print(f"Best Parameters: {grid_lr.best_params_}")
print(f"Best 5-Fold Cross-Validation F1-Score: {grid_lr.best_score_:.4f}")
```

**Observation:**
GridSearchCV tested 8 parameter combinations across 5 stratified folds (40 total fits). Incorporating `class_weight='balanced'` with $C=0.1$ increased the cross-validation F1-score from **0.5930 to 0.6289** (a $+3.59\\%$ absolute gain).

**Conclusion:**
Balancing class weights helps the linear decision boundary overcome class imbalance, significantly boosting minority class sensitivity.

### Task 8.2: RandomizedSearchCV
**Objective:** Explore the expansive hyperparameter space of Random Forest using `RandomizedSearchCV`.
**Method:** Sample configurations over tree count, maximum depth, splitting criteria, leaf sizes, and class weighting over 15 iterations with 5-fold Stratified Cross-Validation optimizing for F1-score.

```python
# Task 8.2: RandomizedSearchCV for Random Forest
pipe_rf_tuning = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(random_state=RANDOM_STATE))
])

param_dist_rf = {
    "classifier__n_estimators": [50, 100, 200],
    "classifier__max_depth": [None, 5, 10, 15],
    "classifier__min_samples_split": [2, 5, 10],
    "classifier__min_samples_leaf": [1, 2, 4],
    "classifier__class_weight": [None, "balanced", "balanced_subsample"]
}

random_rf = RandomizedSearchCV(
    estimator=pipe_rf_tuning,
    param_distributions=param_dist_rf,
    n_iter=15,
    scoring="f1",
    cv=cv_stratified,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

start_time = time.time()
random_rf.fit(X_train, y_train)
rf_tune_time = time.time() - start_time

print(f"RandomizedSearchCV Random Forest completed in {rf_tune_time:.2f}s.")
print(f"Tested Parameter Combinations: 15 iterations across 5 folds (75 total fits)")
print(f"Best Parameters: {random_rf.best_params_}")
print(f"Best 5-Fold Cross-Validation F1-Score: {random_rf.best_score_:.4f}")
```

```python
# Task 11.4: Document Model Input Schema
schema_spec = []
for col in X.columns:
    if col in numerical_features:
        dtype_str = "float64 / int64 (Numerical)"
        allowable = f"[{X[col].min()}, {X[col].max()}]"
    else:
        dtype_str = "object / str (Categorical)"
        allowable = str(sorted(X[col].unique().tolist()))
    schema_spec.append({
        "Feature Column": col,
        "Expected Data Type": dtype_str,
        "Allowable / Domain Values": allowable
    })

display(pd.DataFrame(schema_spec))
```
